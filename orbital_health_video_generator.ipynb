# 🚀 Orbital Health — Automated 1080p Demo Video Generator

This Google Colab notebook automatically launches a headless browser, navigates through **Orbital Health** (https://orbital-health.vercel.app), records real-time 1080p 30/60fps video with all live animations, ECG oscilloscope pulses, and AI interactions, and exports a ready-to-share **MP4 video** for your LinkedIn post!

```python
# Step 2: Automated Interactive Walkthrough & 1080p Video Recording
from playwright.sync_api import sync_playwright
import time, os, glob

os.makedirs('videos', exist_ok=True)

print('🎬 Starting automated browser recording of Orbital Health...')
with sync_playwright() as p:
    browser = p.chromium.launch(
        headless=True,
        args=['--no-sandbox', '--disable-dev-shm-usage', '--disable-gpu']
    )
    context = browser.new_context(
        viewport={'width': 1920, 'height': 1080},
        record_video_dir='videos/',
        record_video_size={'width': 1920, 'height': 1080}
    )
    page = context.new_page()

    # 1. Overview Page & Space Art
    print('1/6: Capturing Overview & Cosmic Backdrop...')
    page.goto('https://orbital-health.vercel.app', wait_until='networkidle', timeout=60000)
    time.sleep(3.5)
    for _ in range(8):
        page.mouse.wheel(0, 320)
        time.sleep(0.45)
    time.sleep(2.0)

    # 2. Live Telemetry Dashboard & Real-Time Lead-II ECG
    print('2/6: Capturing Telemetry Dashboard & 250Hz ECG Oscilloscope...')
    page.click("button[data-view='dashboard']")
    time.sleep(4.5)
    if page.locator('#preset-hypoxia').is_visible():
        page.click('#preset-hypoxia')
        time.sleep(4.0)

    # 3. Interactive Body Map
    print('3/6: Capturing Anatomical Body Map & Organ Systems...')
    page.click("button[data-view='bodymap']")
    time.sleep(3.0)
    if page.locator('#chip-heart').is_visible():
        page.click('#chip-heart')
        time.sleep(2.5)
    if page.locator('#chip-brain').is_visible():
        page.click('#chip-brain')
        time.sleep(2.5)

    # 4. Crew Manifest
    print('4/6: Capturing Crew Manifest & Physiological Roster...')
    page.click("button[data-view='crew']")
    time.sleep(3.5)

    # 5. AURA Medical AI Assistant (Autonomous Flight Surgeon)
    print('5/6: Capturing AURA Medical AI Interface & Telemetry Consultation...')
    page.click("button[data-view='assistant']")
    time.sleep(2.5)
    prompt_buttons = page.locator('.gemini-prompt-card')
    if prompt_buttons.count() > 0:
        prompt_buttons.first.click()
        time.sleep(5.0)

    # 6. Multilateral Space Agency Resources
    print('6/6: Capturing Agency Resources (NASA, CSA, ESA, JAXA)...')
    page.click("button[data-view='sources']")
    time.sleep(4.0)

    # Finalize Video
    context.close()
    browser.close()
    print('✅ Browser walkthrough recorded successfully!')
```

```python
# Step 3: Convert to MP4 (H.264) and Auto-Download to Your Computer
import glob, os
from google.colab import files

video_files = glob.glob('videos/*.webm')
if video_files:
    raw_video = video_files[0]
    output_mp4 = 'orbital_health_demo.mp4'
    print(f'🔄 Converting {raw_video} to {output_mp4} for LinkedIn...')
    !ffmpeg -y -i "{raw_video}" -c:v libx264 -preset fast -crf 20 -pix_fmt yuv420p -r 30 "{output_mp4}"
    print('🎉 Video generated! Download will start automatically in your browser...')
    files.download(output_mp4)
else:
    print('❌ No video file found. Please run Step 2 first.')
```
